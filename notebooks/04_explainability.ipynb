```python
import shap
import joblib
```

```python
shap.initjs()
```

```python
x_test = joblib.load("../data/processed/encoded_x_test.pkl")
best_xgb = joblib.load("../data/processed/best_xgb.pkl")
```

```python
explainer_xgb = shap.TreeExplainer(best_xgb)

shap_values_xgb = explainer_xgb.shap_values(x_test)
```

```python
shap.summary_plot(shap_values_xgb, x_test)
```

```python
shap.summary_plot(shap_values_xgb, x_test, plot_type="bar")
```

```python
i = 0

shap.force_plot(
    explainer_xgb.expected_value,
    shap_values_xgb[i],
    x_test.iloc[i]
)
```
