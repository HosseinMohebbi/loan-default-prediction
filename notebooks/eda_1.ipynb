```python
df.drop(["credit_type", "Security_Type", "construction_type", "Secured_by"], axis=1, inplace=True)
```

```python
df.drop(["Gender", "interest_only"], axis=1, inplace=True)
```

```python
df.drop(["ID", "year"], axis=1, inplace=True)
```

```python
df.columns
```

```python
num_cols = df.select_dtypes(include=["int64", "float64"]).columns.drop("Status")
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
cat_cols = df.select_dtypes(include=["object", "string"]).columns

for col in cat_cols:
    df[col] = df[col].fillna("Unknown")
```

```python
missing = df.isnull().sum().sort_values(ascending=False)
```
